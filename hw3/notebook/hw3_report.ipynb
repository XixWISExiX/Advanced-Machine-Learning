In this report, I will compare 2 Convolutional Neural Network Architectures, one shallow and one deep. To see which one is better using various analysis techniques. These models where trained to classify 4 images (Plug, Scissors, Light Bulb, and Cup) at 128x128 resolution. Also each model was trained with 5 different rotations to avoid outlier results.

## Shared Neural Network Architecture
The Neural Network Architecutre traits that are shared are a dropout of 0.5, spatial_dropout of 0.2, l2 of 0.0001, a convolution activation of relu, and a activation dense function of relu. Output activation function here is softmax.

## Neural Network Architecture Used (Shallow)
For the Shallow NN Architecture I desided to only add one layer to each category.

I had a kernel of size 6, 15 filters, 4 pooling/striding, and 300 hidden units for the dense layer.

```python
from IPython.display import Image

# Display the Image
Image("shallow_NN_Architecture.png", width=600)
```

## Neural Network Architecture Used (Deep)
For the Deep NN Architecture I desided to only add 3 layers to each category.

I had a kernel of size 6, 4, then 3. 10, 30, then 90 filters. 2, 2, and 2 pooling/striding, and 150, 100, and 50 hidden units for the dense layers. Padding was changed to same.

```python
from IPython.display import Image

# Display the Image
Image("deep_NN_Architecture.png", width=600)
```
