## Neural Network Architecture Used
The Neural Network Architecture used for hw0 had only 1 hidden layer and 1024 hidden units (see Figure 1), along with the hidden activation function being ReLu & the output activation function being sigmoid.

The main thoughts for the hidden layer and hidden unit choice was that we didn't have to worry about testing and so here, we could just overfit the data, of which there were 1024 different rows, so I decided to train my model off of 1024 different hidden units to account for each case. Then for the activation function I remembered from class that sigmoid was relatively hard to compute the derivative in comparison to ReLu, and so I also decided to switch the hidden activation function. The reason why I keep the output activation function as sigmoid was because all outputs were binary and sigmoid is commonly used for binary outputs.
